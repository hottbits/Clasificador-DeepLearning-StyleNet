```python
import tensorflow as tf
from tensorflow.keras import layers, models
import matplotlib.pyplot as plt
import numpy as np

# 1. CARGA DE DATOS (Fashion-MNIST) [cite: 258, 287]
fashion_mnist = tf.keras.datasets.fashion_mnist
(train_images, train_labels), (test_images, test_labels) = fashion_mnist.load_data()

# Normalización: los píxeles van de 0 a 1 [cite: 285]
train_images, test_images = train_images / 255.0, test_images / 255.0

# 2. DEFINICIÓN DE LA ARQUITECTURA (CNN) [cite: 257, 290, 293]
model = models.Sequential([
    layers.Conv2D(32, (3, 3), activation='relu', input_shape=(28, 28, 1)),
    layers.MaxPooling2D((2, 2)),
    layers.Conv2D(64, (3, 3), activation='relu'),
    layers.MaxPooling2D((2, 2)),
    layers.Flatten(),
    layers.Dense(128, activation='relu'),
    layers.Dropout(0.2), # Regularización para evitar sobreajuste 
    layers.Dense(10, activation='softmax') # 10 categorías de ropa [cite: 253]
])

# 3. COMPILACIÓN Y ENTRENAMIENTO [cite: 285, 288]
model.compile(optimizer='adam',
              loss='sparse_categorical_crossentropy',
              metrics=['accuracy'])

print("Entrenando el modelo de StyleNet...")
history = model.fit(train_images.reshape(-1,28,28,1), train_labels, epochs=10, 
                    validation_data=(test_images.reshape(-1,28,28,1), test_labels))

# 4. EVALUACIÓN FINAL [cite: 259, 303]
test_loss, test_acc = model.evaluate(test_images.reshape(-1,28,28,1),  test_labels, verbose=2)
print(f'\nExactitud final en prueba: {test_acc:.4f}')
```

```python
# Verificando una predicción real
predictions = model.predict(test_images.reshape(-1,28,28,1))
img_index = 0 # Puedes cambiar este número para ver otras prendas
plt.imshow(test_images[img_index], cmap=plt.cm.binary)
plt.title(f"Predicción: {np.argmax(predictions[img_index])} (Real: {test_labels[img_index]})")
plt.show()
```
